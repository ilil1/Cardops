

## 4. 실행 로드맵
### 역할 분담

| 역할 | 담당 인원 | 주요 업무 |
|---|---|---|
| **데이터/EDA 담당** | 1명 | 데이터 정제, EDA, 데이터 사전 작성, 상관관계 분석 |
| **분류 모델링 담당** | 1~2명 | 로지스틱/RF/XGBoost/LightGBM/MLP 구현 및 튜닝 |

| **해석/평가 담당** | 1명 | SHAP 분석, Baseline 대비 성과(Lift/Gain), A/B 테스트 설계 |
| **서비스화/문서 담당** | 1명 | Streamlit 대시보드, 발표자료, README 취합 |

### 주의할 점
- **"중간 공유"가 핵심**: 각 담당자가 독립적으로 작업하다 보면 전처리 기준(인코딩 방식, 결측치 처리)이 서로 달라질 위험이 있어서, 이 두 시점에 반드시 데이터·모델 산출물을 맞춰야 뒤에서 통합할 때 충돌이 안 나요
- **버전 관리**: GitHub 저장소 하나로 통일하고, 브랜치 전략(예: `data-eda`, `modeling`, `dashboard`)을 미리 정해두면 마지막에 병합할 때 수월해요
- **이탈 정의(라벨 기준)는 1일차에 전원이 합의**해야 함 — 나중에 바뀌면 이미 진행한 모델링/EDA를 다시 해야 하는 리스크가 큼
